```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv("Unemployment in india.csv")
```

```python
df.dropna(inplace=True)
df
```

```python
df = df.sort_values(' Date',ascending=True)
df = df.reset_index(drop=True)
df
```

```python
df.columns
```

```python
df.head(10)
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(subset = ['Region',' Date'],inplace=True)
```

```python
covid_time = pd.to_datetime("2020-03-01")
```

```python
df[' Date'] = pd.to_datetime(df[' Date'],dayfirst=True)
```

```python
df['Period']=np.where(df[" Date"] < covid_time,'Before_covid','After_covid')
df['Period']
```

```python
avg_unemp = df.groupby('Period')[' Estimated Unemployment Rate (%)'].mean()
avg_unemp
```

```python
sns.boxplot(df,x='Period',y=' Estimated Unemployment Rate (%)',palette='Set2')
plt.show()
```

```python
df_trend = df.groupby(' Date')[' Estimated Unemployment Rate (%)'].mean().reset_index()
plt.plot(df_trend[' Date'],df_trend[' Estimated Unemployment Rate (%)'],color = 'b')
```

```python
# Compare average unemployment rate by Area (Urban vs Rural) and Period
area_comparison = df.groupby(['Period', 'Area'])[' Estimated Unemployment Rate (%)'].mean().reset_index()
sns.barplot(data=df, x='Period', y=' Estimated Unemployment Rate (%)', hue='Area', palette='muted')
plt.title('Unemployment Rate by Area: Before vs. After COVID-19')
plt.xlabel('Period')
plt.ylabel('Estimated Unemployment Rate (%)')
plt.legend(title='Area')
plt.show()
```

```python
region_avg = df.groupby('Region')[' Estimated Unemployment Rate (%)'].mean().reset_index()

# 5. Sort the values from highest to lowest for better visualization
region_avg = region_avg.sort_values(by=' Estimated Unemployment Rate (%)', ascending=True)

# 6. Plot the horizontal bar chart
plt.figure(figsize=(12, 8))
plt.barh(region_avg['Region'], region_avg[' Estimated Unemployment Rate (%)'], color='teal')

# Add titles and labels
plt.title('Average Unemployment Rate by Region (State-wise)', fontsize=14)
plt.xlabel('Average Estimated Unemployment Rate (%)', fontsize=12)
plt.ylabel('Region', fontsize=12)
```
